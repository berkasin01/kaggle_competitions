```python
import pandas as pd 
import numpy as np
from matplotlib import pyplot
```

```python
train_df = pd.read_csv("dataset/train.csv")
test_df = pd.read_csv("dataset/test.csv")
```

```python
train_df_copy = train_df.copy()
y = train_df_copy["SalePrice"].values
train_df_copy = train_df_copy.drop(["SalePrice"],axis=1)
```

```python
train_df["PoolQC"].value_counts()
```

```python
cols = train_df_copy.columns.to_list()
print(train_df_copy.shape)
for c in cols:
    curr_col = train_df_copy[c]
    curr_na_vals = curr_col.isna().sum()
    if(curr_na_vals > 100):
        print(f"col:{c},na:{curr_na_vals}")
```

```python
train_df_copy = train_df_copy.drop(["Alley","PoolQC","Fence","MiscFeature"],axis=1)
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.impute import SimpleImputer

na_transformers = {}
na_cols = ["LotFrontage","MasVnrType","FireplaceQu"]

for x in na_cols:
    curr = train_df_copy[x]
    print(x)
    print(len(curr.value_counts()))
    print(curr.dtype)
    
    if(curr.dtype == "str"):
        le = LabelEncoder()
        le.fit(curr)
        na_transformers[x] = le
        train_df_copy[x] = le.transform(train_df_copy[x])
    elif(curr.dtype == "float64"):
        imputer = SimpleImputer(strategy="mean")
        imputer.fit(train_df_copy[[x]])
        na_transformers[x] = imputer
        train_df_copy[x] = imputer.transform(train_df_copy[[x]])
```

```python
train_df_copy["LotFrontage"].value_counts()
```

```python
train_df_copy["FireplaceQu"].value_counts()
```

```python
from sklearn.preprocessing import StandardScaler

all_cols = train_df_copy.columns.to_list()
str_cols = []
flt_cols = []
int_cols = []

col_encoders = {}

continuous_int_cols = ['LotArea', 'BsmtFinSF1', 'BsmtFinSF2', 'BsmtUnfSF', 
           'TotalBsmtSF', '1stFlrSF', '2ndFlrSF', 'GrLivArea', 
           'GarageArea', 'WoodDeckSF', 'OpenPorchSF', 'EnclosedPorch', 
           'ScreenPorch', 'PoolArea', 'MiscVal', 'LowQualFinSF', '3SsnPorch']

for i in all_cols:
    curr_col = train_df_copy[i]
    if curr_col.dtype == "str":
        le = LabelEncoder()
        le.fit(curr_col)
        col_encoders[i] = le
        train_df_copy[i] = le.transform(train_df_copy[i])
        # print(curr_col.value_counts())
        str_cols.append(i)
    elif curr_col.dtype == "float64":
        ss = StandardScaler()
        ss.fit(train_df_copy[[i]])
        col_encoders[i] = ss
        train_df_copy[i] = ss.transform(train_df_copy[[i]])
        
        flt_cols.append(i)
    elif curr_col.dtype == "int64":
        if i in continuous_int_cols:
            ss = StandardScaler()
            ss.fit(train_df_copy[[i]])
            col_encoders[i] = ss
            train_df_copy[i] = ss.transform(train_df_copy[[i]])
            
        int_cols.append(i)        
```

```python
X = train_df_copy
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.1,random_state=42)
```

```python
from xgboost import XGBRegressor

model = XGBRegressor(n_estimators=1000, learning_rate=0.05, random_state=42)
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np
from sklearn.metrics import r2_score


y_hat = model.predict(X_test)
r2 = r2_score(y_test, y_hat)
rmse = mean_squared_error(y_test,y_hat)
mse = np.sqrt(rmse)
```

```python
mse,r2
```

```python
train_df["SalePrice"].mean()
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'n_estimators': [500, 1000, 1500],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.05, 0.1],
    'subsample': [0.7, 0.8, 1.0]
}

param_model = XGBRegressor()

grid_search = GridSearchCV(param_model, param_grid, cv=5, scoring="neg_mean_squared_error",return_train_score=True)
grid_search.fit(X_train, y_train)
```

```python
grid_search.best_params_
```

```python
best_model = grid_search.best_estimator_
```

```python
best_y_hat = best_model.predict(X_test)
r2 = r2_score(y_test, best_y_hat)
rmse = mean_squared_error(y_test,best_y_hat)
mse = np.sqrt(rmse)
```

```python
test_df_copy = test_df.copy()
test_df_copy = test_df_copy.drop(["Alley","PoolQC","Fence","MiscFeature"],axis=1)
```

```python
for i in str_cols:
    if "missing" not in col_encoders[i].classes_:
        col_encoders[i].classes_ = np.append(col_encoders[i].classes_, "missing")

# Then transform
for i in all_cols:
    if i in col_encoders:
        curr_col = test_df_copy[i]
        if curr_col.dtype == "str":
            test_df_copy[i] = test_df_copy[i].fillna("missing")
            test_df_copy[i] = col_encoders[i].transform(test_df_copy[i])
        else:
            test_df_copy[i] = col_encoders[i].transform(test_df_copy[[i]])
```

```python
final_model = XGBRegressor(learning_rate=0.05, max_depth=3, n_estimators=1500, subsample=0.8)
final_model.fit(X, y)
final_preds = final_model.predict(test_df_copy)
```
